### NSE Daily close of NIFTY index

```python
import requests
import pandas as pd
from io import StringIO
```

```python
url = "https://www.nseindia.com/api/historicalOR/indicesHistory"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json",
    "Referer": "https://www.nseindia.com/reports-indices-historical-index-data",
}

params = {
    "indexType": "NIFTY 50",
    "from": "01-01-2026",
    "to": "31-12-2026"
}

session = requests.Session()
session.headers.update(headers)

# Establish NSE session first
session.get("https://www.nseindia.com", timeout=20)

response = session.get(
    url,
    params=params,
    timeout=30
)

response.raise_for_status()

data = response.json()

data.keys()
```

```python
df = pd.DataFrame(data["data"])
```

```python
print(data["data"][0])
print(data["data"][-1])
```

```python
# params = {
#     "indexType": "NIFTY 50",
#     "from": "01-01-2026",
#     "to": "31-01-2026"
# }

# params = {
#     "indexType": "NIFTY 50",
#     "from": "01-01-2026",
#     "to": "31-03-2026"
# }

params1 = {
    "indexType": "NIFTY 50",
    "from": "01-01-2026",
    "to": "31-03-2026"
}

response1 = session.get(
    url,
    params=params1,
    timeout=30
)

data1 = response1.json()["data"]

params2 = {
    "indexType": "NIFTY 50",
    "from": "01-04-2026",
    "to": "30-06-2026"
}

response2 = session.get(
    url,
    params=params2,
    timeout=30
)

data2 = response2.json()["data"]

print(len(data1))
print(data1[0]["EOD_TIMESTAMP"])
print(data1[-1]["EOD_TIMESTAMP"])

print(len(data2))
print(data2[0]["EOD_TIMESTAMP"])
print(data2[-1]["EOD_TIMESTAMP"])
```

```python
nifty1 = pd.DataFrame([
    {
        "date": x["EOD_TIMESTAMP"],
        "close": x["EOD_CLOSE_INDEX_VAL"]
    }
    for x in data1
])

nifty1["date"] = pd.to_datetime(nifty1["date"],format="%d-%b-%Y")

nifty2 = pd.DataFrame([
    {
        "date": x["EOD_TIMESTAMP"],
        "close": x["EOD_CLOSE_INDEX_VAL"]
    }
    for x in data2
])

nifty2["date"] = pd.to_datetime(nifty2["date"],format="%d-%b-%Y")

nifty1 = nifty1.sort_values("date").reset_index(drop=True)
nifty2 = nifty2.sort_values("date").reset_index(drop=True)
nifty = pd.concat([nifty1, nifty2], axis = 0)
```

```python
nifty.to_csv('nift_close.csv', index= False)
```
