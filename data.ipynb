```python
import pandas as pd
from huggingface_hub import hf_hub_download
import duckdb
```

```python
file_path = hf_hub_download(
    repo_id="rissin/nse-options-intraday",
    filename="upstox_intraday/NIFTY/NIFTY_2026.parquet",
    repo_type="dataset"
)

df = pd.read_parquet(file_path)
```
